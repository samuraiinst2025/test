```python
# 1. データ分析用ライブラリを読み込み
import pandas as pd

# 2. データを読み込んで統合する関数を定義
def load_order_data(file_paths):
    # 空の表（データフレーム）を作成
    orders = pd.DataFrame()

    # 住所リスト（file_paths）を1つずつ取り出して処理
    for file_path in file_paths:
        # Excelファイルを読み込む
        order = pd.read_excel(file_path, engine='openpyxl')
        # 既存の表（orders）の下に新しい表（order）を結合
        orders = pd.concat([orders, order], ignore_index=True)

    # 完成した大きな表を外に返す
    return orders

# 3. 実際に統合を実行！
all_orders = load_order_data(file_paths)

# 4. 結合結果を表示して確認
print(f"✅ 全店舗の注文を統合しました。行数・列数:")
print(all_orders.shape)
print(all_orders.head()) # 最初の5行だけ表示

```

```python
import datetime
import smtplib
from email.mime.text import MIMEText
from email.header import Header
import pandas as pd

def run_order_system():
    print("🚀 自動発注・在庫更新システムを開始します...")

    # -----------------------------------------------
    # 1. 各店から届く注文情報をまとめる（要件1）
    # -----------------------------------------------
    vegetable_columns = ['トマト', 'キャベツ', 'レタス', '白菜', 'ほうれん草', '大根', 'ニンジン']
    all_orders = load_order_data(file_paths)  # 以前定義した注文読み込み関数
    total_orders = all_orders[vegetable_columns].sum()

    # -----------------------------------------------
    # 2. 現在の在庫状況を確認する（要件2）
    # -----------------------------------------------
    inventory_df = pd.read_excel('/content/drive/MyDrive/samples/inventory.xlsx', engine='openpyxl')
    # 最終行の在庫情報を抽出
    inventory_series = inventory_df[vegetable_columns].iloc[-1]

    # -----------------------------------------------
    # 3. 発注が必要なアイテムを特定する（要件3）
    # -----------------------------------------------
    # 「最新の在庫」-「合計注文数」
    remaining_stock = inventory_series - total_orders

    # pickup.xlsx からしきい値を確認
    pickup = pd.read_excel('/content/drive/MyDrive/samples/pickup.xlsx', engine='openpyxl')
    thresholds_initial = pd.Series(data=pickup.loc[0, vegetable_columns].values, index=vegetable_columns)
    thresholds = thresholds_initial.reindex(remaining_stock.index).astype(float)

    # しきい値を下回った野菜のみ特定
    order_list = remaining_stock[remaining_stock < thresholds]

    # -----------------------------------------------
    # 4. 発注のためのメール文を作成・送信（要件4・5）
    # -----------------------------------------------
    if not order_list.empty:
        print("⚠️ しきい値を下回った商品があります。発注メールを作成します...")

        # メール本文の組み立て（不足数はプラスに反転）
        email_body = "農家様\n\nお世話になっております。在庫不足のため、以下の追加発注をお願いします。\n\n"
        for vegetable, quantity in (order_list * -1).items():
            email_body += f"・{vegetable} : {int(quantity)}個\n"
        email_body += "\nよろしくお願いいたします。"

        # Mailtrapへ送信
        SMTP_HOST = "sandbox.smtp.mailtrap.io"
        SMTP_PORT = 2525
        SMTP_USER = "8d5866beac0a0"
        SMTP_PASS = "fd3e5126613c3c"  # 実績のあったパスワード

        msg = MIMEText(email_body, "plain", "utf-8")
        msg["Subject"] = Header("【自動発注】野菜在庫不足による追加発注の御願い", "utf-8")
        msg["From"] = "system@example.com"
        msg["To"] = "nouka@example.com"

        try:
            with smtplib.SMTP(SMTP_HOST, SMTP_PORT) as server:
                server.starttls()
                server.login(SMTP_USER, SMTP_PASS)
                server.send_message(msg)
            print("📨 Mailtrapへ発注メールを正常に送信しました！")
        except Exception as e:
            print(f"❌ メール送信エラー: {e}")
    else:
        print("✅ すべての野菜がしきい値以上です。発注メールの送信はありません。")

    # -----------------------------------------------
    # 5. 在庫状況を最新の情報に更新して上書き（要件6）
    # -----------------------------------------------
    # 新しい在庫行を作成（日付 + 計算した残り在庫）
    new_row = remaining_stock.to_dict()
    new_row['日付'] = datetime.date.today().strftime('%Y-%m-%d')

    # inventory.xlsxの最下行へ追加して保存
    updated_inventory_df = pd.concat([inventory_df, pd.DataFrame([new_row])], ignore_index=True)
    updated_inventory_df.to_excel('/content/drive/MyDrive/samples/inventory.xlsx', index=False)
    print("💾 inventory.xlsx の最下行に最新在庫を追加し、上書き保存しました。")
    print("✨ すべての連続処理が正常に終了しました。")

# システムを一発実行！
run_order_system()
```
